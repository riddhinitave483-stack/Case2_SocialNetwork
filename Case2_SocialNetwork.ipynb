```python

# Case Study 2: Social Network using Adjacency Matrix

# Step 1: Take number of users
n = int(input("Enter number of users: "))

# Step 2: Take user names
users = []

for i in range(n):
    name = input(f"Enter name of user {i + 1}: ")
    users.append(name)

# Step 3: Create adjacency matrix
matrix = [[0 for j in range(n)] for i in range(n)]

# Step 4: Take friendship connections
print("\nEnter friendship connections.")
print("Enter the numbers of two users to make them friends.")
print("User numbers are from 1 to", n)
print("Enter -1 to stop adding friendships.")

while True:
    a = int(input("\nEnter first user number: "))

    if a == -1:
        break

    b = int(input("Enter second user number: "))

    if a < 1 or a > n or b < 1 or b > n:
        print("Invalid user number! Try again.")
        continue

    if a == b:
        print("A user cannot be friends with themselves.")
        continue

    # Convert user numbers to matrix indexes
    a -= 1
    b -= 1

    # Friendship is mutual
    matrix[a][b] = 1
    matrix[b][a] = 1

    print("Friendship added successfully!")

# Step 5: Display adjacency matrix
print("\n--- Social Network Adjacency Matrix ---")

print(" " * 12, end="")
for name in users:
    print(f"{name:>10}", end="")
print()

for i in range(n):
    print(f"{users[i]:>12}", end="")
    for j in range(n):
        print(f"{matrix[i][j]:>10}", end="")
    print()

# Step 6: Display friends of each user
print("\n--- Friends List ---")

for i in range(n):
    friends = []

    for j in range(n):
        if matrix[i][j] == 1:
            friends.append(users[j])

    print(f"{users[i]} has {len(friends)} friend(s):", end=" ")

    if friends:
        print(", ".join(friends))
    else:
        print("No friends")

print("\nSocial Network program completed!")
```
